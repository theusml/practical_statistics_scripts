```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
rng = np.random.default_rng(42)
sample_data = rng.normal(loc=62_231, scale=5_000, size=20)
sample_data
```

```python
def bootstrap_resample(data: np.ndarray, rng: np.random.Generator) -> np.ndarray:
    """Return one bootstrap sample with the same size as the input."""
    return rng.choice(data, size=len(data), replace=True)


bootstrap_samples = np.array(
    [bootstrap_resample(sample_data, rng) for _ in range(10_000)]
)
sample_statistics = bootstrap_samples.mean(axis=1)
sample_mean = sample_data.mean()
```

```python
confidence_level = 95
lower_bound, upper_bound = np.percentile(
    sample_statistics,
    [(100 - confidence_level) / 2, 100 - (100 - confidence_level) / 2],
)

plt.figure(figsize=(6, 4))
sns.histplot(sample_statistics, kde=True, color="grey")
plt.axvline(
    lower_bound,
    color="red",
    linestyle="dashed",
    label=f"Limite inferior ({lower_bound:,.2f})",
)
plt.axvline(
    upper_bound,
    color="red",
    linestyle="dashed",
    label=f"Limite superior ({upper_bound:,.2f})",
)
plt.axvline(
    sample_mean,
    color="blue",
    linestyle="dashed",
    label=f"Média ({sample_mean:,.2f})",
)
plt.title(f"Distribuição bootstrap com IC de {confidence_level}%")
plt.xlabel("Média das amostras reamostradas")
plt.ylabel("Frequência")
plt.legend()
plt.show()
```
